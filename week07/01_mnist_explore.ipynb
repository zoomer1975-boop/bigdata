## 목표

- 정답 레이블의 분포를 본다

2주차에서 "이미지는 숫자 배열"이라고 했던 것을 실제 데이터로 확인합니다.


```python
import torch
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

torch.manual_seed(0)
```

## 1. 내려받기
`ToTensor()` 는 0~255 픽셀을 0~1 실수로 바꾸고 (채널, 높이, 너비) 모양의 Tensor로 만듭니다.


```python
transform = transforms.ToTensor()

train_set = datasets.MNIST(root="./data", train=True,  download=True, transform=transform)
test_set  = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

print("학습 데이터:", len(train_set))
print("시험 데이터:", len(test_set))
```

## 2. 한 장의 구조


```python
img, label = train_set[0]

print("타입    :", type(img))
print("shape   :", img.shape)          # (1, 28, 28)  채널 1개
print("dtype   :", img.dtype)
print("최솟값  :", img.min().item())
print("최댓값  :", img.max().item())
print("정답    :", label)
```

```python
# 픽셀 값을 숫자로 직접 보기 — 가운데 부분만
print(img[0, 10:18, 10:18].round(decimals=2))
```

## 3. 이미지로 그려 보기


```python
plt.figure(figsize=(3, 3))
plt.imshow(img.squeeze(), cmap="gray")
plt.title(f"label = {label}")
plt.axis("off")
plt.show()
```

```python
# 여러 장 한 번에
fig, axes = plt.subplots(2, 8, figsize=(14, 4))
for ax, i in zip(axes.flatten(), range(16)):
    im, lb = train_set[i]
    ax.imshow(im.squeeze(), cmap="gray")
    ax.set_title(str(lb), fontsize=11)
    ax.axis("off")
plt.tight_layout(); plt.show()
```

## 4. 정답 레이블의 분포
클래스가 크게 치우쳐 있으면 정확도만 보고 판단할 수 없습니다. MNIST는 비교적 고른 편입니다.


```python
labels = train_set.targets
print("클래스:", labels.unique().tolist())

counts = torch.bincount(labels)
for d, c in enumerate(counts.tolist()):
    print(f"  {d}: {c:5d}개")

plt.bar(range(10), counts)
plt.xticks(range(10)); plt.xlabel("digit"); plt.ylabel("count")
plt.title("label distribution"); plt.show()
```
