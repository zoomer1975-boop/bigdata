```python
import torch
print("torch 버전:", torch.__version__)
```

## 1. Tensor 만들기


```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.zeros(2, 3)
c = torch.ones(2, 3)
d = torch.arange(0, 10, 2)
e = torch.rand(2, 3)          # 0~1 균등분포
f = torch.randn(2, 3)         # 표준정규분포

print(a)
print(b)
print(d)
print(f)
```

## 2. shape 과 dtype — 딥러닝 오류의 90%는 이 두 가지 때문입니다


```python
x = torch.randn(4, 3)

print("shape:", x.shape)      # (행, 열)
print("차원 수:", x.dim())
print("원소 수:", x.numel())
print("dtype:", x.dtype)
print("device:", x.device)
```

```python
# 모양 바꾸기
y = x.reshape(3, 4)
z = x.view(-1)                # -1 은 "알아서 계산"
print(y.shape, z.shape)

# 행/열 추가 — 배치 차원을 만들 때 자주 씁니다
print(x.unsqueeze(0).shape)   # (1, 4, 3)
```

## 3. 사칙연산과 행렬곱


```python
p = torch.tensor([[1.0, 2.0],
                  [3.0, 4.0]])
q = torch.tensor([[10.0, 20.0],
                  [30.0, 40.0]])

print(p + q)          # 원소별 덧셈
print(p * q)          # 원소별 곱셈 (행렬곱이 아닙니다)
print(p @ q)          # 행렬곱
print(p.mean(), p.sum(), p.max())
```

## 4. NumPy 와 오가기
통계학에서 쓰던 배열을 그대로 가져올 수 있습니다.


```python
import numpy as np

arr = np.array([[1.0, 2.0], [3.0, 4.0]])
t = torch.from_numpy(arr)     # NumPy -> Tensor
back = t.numpy()              # Tensor -> NumPy

print(type(arr), type(t), type(back))
print(t)
```

## 5. GPU로 옮기기
`런타임 > 런타임 유형 변경 > T4 GPU` 를 먼저 선택하세요.


```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("사용할 장치:", device)

x_gpu = x.to(device)
print(x_gpu.device)
```

```python
# CPU 텐서와 GPU 텐서는 섞어서 연산할 수 없습니다 — 흔한 오류
try:
    x_gpu + torch.randn(4, 3)
except RuntimeError as err:
    print("RuntimeError:", err)
```

## 6. 자동미분 맛보기
`requires_grad=True` 를 주면 PyTorch가 미분을 대신 계산합니다.
$y = x^2$ 에서 $x=3$ 일 때 도함수는 $2x = 6$ 입니다.


```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2
y.backward()
print("dy/dx =", x.grad)
```
