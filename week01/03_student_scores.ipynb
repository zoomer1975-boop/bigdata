## 요구사항
1. 학생 이름과 점수를 함께 저장한다
2. 전체 평균을 소수점 둘째 자리까지 출력한다
3. 최고 점수 학생의 이름과 점수를 출력한다
4. 70점 이상이면 합격, 미만이면 불합격을 학생별로 출력한다
5. 합격률을 백분율로 출력한다

먼저 **직접 작성**해 보고, 아래 모범 답안과 비교하세요.


```python
students = {
    "김통계": 88,
    "이확률": 92,
    "박회귀": 79,
    "최추정": 95,
    "정검정": 61,
}

# 여기에 작성하세요
```

---
## 모범 답안
아래 셀은 스스로 작성한 뒤에 실행하세요.


```python
def report(students, cutoff=70):
    scores = list(students.values())
    mean = sum(scores) / len(scores)
    top_name = max(students, key=students.get)

    print(f"전체 평균 : {mean:.2f}")
    print(f"최고 점수 : {top_name} ({students[top_name]}점)")
    print("-" * 28)

    n_pass = 0
    for name, score in students.items():
        if score >= cutoff:
            result = "합격"
            n_pass += 1
        else:
            result = "불합격"
        print(f"{name:5s} {score:3d}점  {result}")

    print("-" * 28)
    print(f"합격률 : {n_pass / len(students) * 100:.1f}%")


report(students)
```
