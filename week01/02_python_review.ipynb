## 3. 리스트


```python
scores = [88, 92, 79, 95, 61]

print("개수:", len(scores))
print("첫 번째:", scores[0], "/ 마지막:", scores[-1])
print("앞 세 개:", scores[:3])
print("합계:", sum(scores))
print("평균:", sum(scores) / len(scores))
print("최고:", max(scores), "/ 최저:", min(scores))
print("정렬:", sorted(scores, reverse=True))
```

```python
# 리스트 컴프리헨션 — NumPy/PyTorch 코드에서 자주 등장합니다
squared = [s ** 2 for s in scores]
high = [s for s in scores if s >= 90]
print(squared)
print(high)
```

## 4. 딕셔너리


```python
student = {"name": "김통계", "score": 87.5, "major": "통계학과"}

print(student["name"])
print(student.get("grade", "없음"))   # 없는 키를 안전하게 읽기

student["grade"] = "B+"
for key, value in student.items():
    print(key, "->", value)
```

## 5. 조건문과 반복문


```python
for s in scores:
    if s >= 90:
        grade = "A"
    elif s >= 80:
        grade = "B"
    elif s >= 70:
        grade = "C"
    else:
        grade = "F"
    print(f"{s}점 -> {grade}")
```

```python
# enumerate: 순번과 값을 함께
for i, s in enumerate(scores, start=1):
    print(i, "번째 학생:", s)

# zip: 두 리스트를 짝지어
names = ["김통계", "이확률", "박회귀", "최추정", "정검정"]
for n, s in zip(names, scores):
    print(n, s)
```

## 6. 함수와 반환값


```python
def summarize(values):
    """리스트를 받아 평균, 최고점, 합격 인원을 반환한다."""
    mean = sum(values) / len(values)
    best = max(values)
    n_pass = len([v for v in values if v >= 70])
    return mean, best, n_pass


mean, best, n_pass = summarize(scores)
print(f"평균 {mean:.2f} / 최고 {best} / 합격 {n_pass}명")
```

```python
# 기본값이 있는 매개변수
def is_pass(score, cutoff=70):
    return score >= cutoff

print(is_pass(68))
print(is_pass(68, cutoff=60))
```

## 7. 파일 입출력 (CSV)
Colab에는 파일이 없으므로, 먼저 CSV를 만들고 그 파일을 읽습니다.


```python
csv_text = """name,score
김통계,88
이확률,92
박회귀,79
최추정,95
정검정,61
"""

with open("scores.csv", "w", encoding="utf-8") as f:
    f.write(csv_text)

print("파일을 만들었습니다.")
```

```python
# 방법 1 — 표준 라이브러리
import csv

with open("scores.csv", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

for r in rows:
    print(r["name"], int(r["score"]))
```

```python
# 방법 2 — pandas (2주차부터 본격적으로 사용합니다)
import pandas as pd

df = pd.read_csv("scores.csv")
print(df)
print("\n평균:", df["score"].mean())
```
