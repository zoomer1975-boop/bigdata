## 1. NameError — 만들지 않은 이름을 썼다


```python
print(total_score)
```

```python
# 해결: 변수를 먼저 만든다
total_score = 415
print(total_score)
```

## 2. TypeError — 서로 맞지 않는 자료형을 섞었다


```python
age = 22
print("나이: " + age)
```

```python
# 해결 1: 형 변환
print("나이: " + str(age))
# 해결 2: f-string
print(f"나이: {age}")
```

## 3. IndexError — 없는 위치를 꺼냈다


```python
scores = [88, 92, 79]
print(scores[3])
```

```python
# 해결: 인덱스는 0부터 시작한다. 마지막은 len(scores) - 1
print(scores[2])
print(scores[-1])
```

## 4. IndentationError — 들여쓰기가 어긋났다


```python
for s in scores:
print(s)
```

## 5. ValueError / KeyError — 실습 중 자주 만나는 두 가지


```python
int("점수")
```

```python
student = {"name": "김통계"}
print(student["score"])
```

```python
# 해결: get 으로 안전하게 읽는다
print(student.get("score", "점수 없음"))
```
