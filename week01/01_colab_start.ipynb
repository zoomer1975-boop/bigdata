## 3. 현재 환경 확인
Colab은 구글 서버에서 실행되는 리눅스 컴퓨터입니다.


```python
import sys, platform
print("파이썬 버전:", sys.version.split()[0])
print("운영체제:", platform.platform())
```

```python
# GPU가 연결되어 있는지 확인 (런타임 > 런타임 유형 변경 > T4 GPU)
import torch
print("torch 버전:", torch.__version__)
print("GPU 사용 가능:", torch.cuda.is_available())
```
