## 계획을 코드로 확인하기
계획한 shape이 실제로 맞는지 여기서 검증합니다. 학습 전에 반드시 하세요.


```python
import torch
import torch.nn as nn

# 계획서에 적은 값을 여기에 옮겨 적으세요
BATCH = 32
INPUT_SHAPE = (1, 28, 28)      # 채널, 높이, 너비 (이미지) 또는 (시점, 특성) (시계열)
N_CLASSES = 10

dummy = torch.randn(BATCH, *INPUT_SHAPE)
print("계획한 입력:", tuple(dummy.shape))
```

```python
# 모델을 여기에 작성하세요
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(28*28, 128), nn.ReLU(),
    nn.Linear(128, N_CLASSES),
)

out = model(dummy)
print("출력:", tuple(out.shape))
assert out.shape == (BATCH, N_CLASSES), "출력 shape이 계획과 다릅니다"
print("→ 계획한 shape과 일치합니다")
print("파라미터:", f"{sum(p.numel() for p in model.parameters()):,}개")
```

```python
# 손실 함수도 확인
loss_fn = nn.CrossEntropyLoss()
y = torch.randint(0, N_CLASSES, (BATCH,))
print("정답 shape:", tuple(y.shape), "dtype:", y.dtype)
print("손실:", loss_fn(out, y).item())
```

## 성공 기준의 하한선 확인
무작위로 찍으면 몇 점인지 먼저 알아야 합니다.


```python
print(f"클래스 {N_CLASSES}개 · 무작위 추측 정확도 = {1/N_CLASSES:.4f}")
print("→ 이보다 못하면 학습이 안 된 것입니다")

# 학습 전 모델의 정확도 — 무작위와 비슷해야 정상
with torch.no_grad():
    pred = model(dummy).argmax(dim=1)
print(f"학습 전 정확도: {(pred == y).float().mean():.4f}")
```
