## 목표
- 가중치 곱 → 편향 더하기 → 활성화 함수의 순서를 확인한다
- 직접 계산한 값과 `nn.Linear` 의 결과를 비교한다
- 활성화 함수의 모양을 그려 본다

뉴런 하나가 하는 일은 `활성화(입력 · 가중치 + 편향)` 뿐입니다.


```python
import torch
import torch.nn as nn

torch.manual_seed(0)
```

## 1. 손으로 계산하기
입력 3개, 출력 1개인 뉴런입니다.


```python
x = torch.tensor([2.0, 3.0, 1.0])          # 입력
w = torch.tensor([0.5, -1.0, 2.0])         # 가중치
b = torch.tensor(1.0)                       # 편향

z = (x * w).sum() + b                       # 가중합
print("가중합 z =", z.item())

out = torch.relu(z)                         # 활성화
print("ReLU(z) =", out.item())
```

```python
# 행렬곱으로 쓰면 이렇게 됩니다 — 4주차의 @ 연산
z2 = x @ w + b
print(z2.item())
```

## 2. nn.Linear 와 비교
`nn.Linear` 는 가중치와 편향을 가진 층입니다. 같은 값을 넣어 확인합니다.


```python
layer = nn.Linear(3, 1)

# 가중치를 위에서 쓴 값으로 직접 지정
with torch.no_grad():
    layer.weight.copy_(w.reshape(1, 3))
    layer.bias.copy_(b.reshape(1))

print("weight:", layer.weight.data)
print("bias  :", layer.bias.data)
print("결과  :", layer(x).item())            # 위와 같은 값
```

## 3. 활성화 함수가 없으면
층을 여러 개 쌓아도 결국 하나의 선형식이 됩니다. 다음 노트북에서 확인합니다.


```python
print("ReLU   :", torch.relu(torch.tensor([-2.0, -0.5, 0.0, 0.5, 2.0])))
print("Sigmoid:", torch.sigmoid(torch.tensor([-2.0, 0.0, 2.0])))
print("Tanh   :", torch.tanh(torch.tensor([-2.0, 0.0, 2.0])))
```

```python
import matplotlib.pyplot as plt

t = torch.linspace(-5, 5, 200)

fig, ax = plt.subplots(1, 3, figsize=(12, 3))
ax[0].plot(t, torch.relu(t));    ax[0].set_title("ReLU");    ax[0].grid(alpha=.3)
ax[1].plot(t, torch.sigmoid(t)); ax[1].set_title("Sigmoid"); ax[1].grid(alpha=.3)
ax[2].plot(t, torch.tanh(t));    ax[2].set_title("Tanh");    ax[2].grid(alpha=.3)
plt.tight_layout(); plt.show()
```

## 4. 뉴런 여러 개 — 출력이 여러 개인 층
`nn.Linear(3, 4)` 는 뉴런 4개입니다. 가중치는 (4, 3) 행렬이 됩니다.


```python
layer = nn.Linear(3, 4)

print("weight shape:", layer.weight.shape)   # (출력, 입력)
print("bias shape  :", layer.bias.shape)
print("파라미터 개수:", sum(p.numel() for p in layer.parameters()))   # 4*3 + 4

out = layer(x)
print("출력:", out.shape, out.data.round(decimals=3))
```

## 5. batch로 넣기
5주차의 batch가 그대로 들어갑니다. 첫 번째 축만 batch 크기입니다.


```python
batch = torch.randn(8, 3)          # 데이터 8개
print("입력:", batch.shape)
print("출력:", layer(batch).shape)   # (8, 4)
```
