## 목표
- 원소별 연산과 행렬곱을 구분한다
- 평균 · 합계 · 최댓값을 축 방향으로 계산한다
- 브로드캐스팅 규칙을 확인한다


```python
import torch

p = torch.tensor([[1.0, 2.0],
                  [3.0, 4.0]])
q = torch.tensor([[10.0, 20.0],
                  [30.0, 40.0]])
print(p)
print(q)
```

## 1. 원소별 연산 — 같은 위치끼리


```python
print(p + q)
print(p - q)
print(p * q)          # 원소별 곱 (행렬곱이 아닙니다)
print(p / q)
print(p ** 2)
```

## 2. 행렬곱 — @ 또는 matmul
`(m, n) @ (n, p)` 에서 가운데 n이 같아야 합니다.


```python
print(p @ q)
print(torch.matmul(p, q))
print((p * q).shape, (p @ q).shape)
```

```python
# 차원이 맞지 않으면 오류
a = torch.randn(2, 3)
b = torch.randn(2, 3)
try:
    a @ b
except RuntimeError as err:
    print("RuntimeError:", err)

print("해결:", (a @ b.T).shape)
```

## 3. 통계 연산 — 축을 지정합니다
2주차 NumPy의 axis와 같은 개념입니다. `dim` 이 사라지는 방향입니다.


```python
x = torch.tensor([[88., 92., 79.],
                  [95., 61., 84.],
                  [70., 100., 66.],
                  [82., 74., 91.]])

print("전체 평균 :", x.mean())
print("열별 평균 :", x.mean(dim=0), x.mean(dim=0).shape)
print("행별 평균 :", x.mean(dim=1), x.mean(dim=1).shape)
```

```python
print("합계  :", x.sum(dim=0))
print("최댓값:", x.max(dim=1).values)
print("위치  :", x.max(dim=1).indices)
print("argmax:", x.argmax(dim=1))
print("표준편차:", x.std(dim=0))
```

```python
# keepdim — 축을 남겨 두면 브로드캐스팅에 편합니다
print(x.mean(dim=1).shape)                  # (4,)
print(x.mean(dim=1, keepdim=True).shape)    # (4, 1)

# keppdim 형태 확인
y = x.mean(dim=1, keepdim=True)
print(y)

# 행별 평균을 빼기
centered = x - x.mean(dim=1, keepdim=True)
#print(centered.round())
```

## 4. 브로드캐스팅


```python
print(x + 10)                               # 스칼라
print(x + torch.tensor([1., 2., 3.]))       # (4,3) + (3,)
print((x + torch.tensor([[1.], [2.], [3.], [4.]])).shape)   # (4,3) + (4,1)
```

```python
# 표준화 — 열 기준
z = (x - x.mean(dim=0)) / x.std(dim=0)
print(z.round(decimals=2))
print("열별 평균:", z.mean(dim=0).round(decimals=4))
```

## 5. 값 하나 꺼내기 — item()


```python
total = x.sum()
print(total)               # Tensor
print(total.item())       # 파이썬 실수 — print나 기록에 쓸 때
```
