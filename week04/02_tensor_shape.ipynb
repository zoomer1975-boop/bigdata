## 목표
- `shape` `size()` `dim()` `dtype` `device` 를 읽는다
- `reshape` `view` `squeeze` `unsqueeze` 로 모양을 바꾼다
- Tensor 속성 확인표를 만든다


```python
import torch

x = torch.randn(4, 3)
x
```

## 1. 다섯 가지 속성 — 막히면 이것부터 출력


```python
print("shape  :", x.shape)
print("size() :", x.size())        # shape 과 같은 값
print("dim()  :", x.dim())         # 차원 수
print("dtype  :", x.dtype)
print("device :", x.device)
print("numel  :", x.numel())       # 전체 원소 수
```

```python
# 특정 축의 크기만
print(x.shape[0], x.size(1))
```

## 2. Tensor 속성 확인표 만들기
여러 Tensor를 한 표로 정리하면 오류를 훨씬 빨리 찾습니다.


```python
import pandas as pd

tensors = {
    "a 정수":   torch.tensor([1, 2, 3]),
    "b 실수":   torch.tensor([1.0, 2.0, 3.0]),
    "c 영행렬": torch.zeros(2, 3),
    "d 난수":   torch.randn(2, 3, 4),
    "e 스칼라": torch.tensor(7.0),
}

rows = []
for name, t in tensors.items():
    rows.append({"이름": name, "shape": tuple(t.shape), "dim": t.dim(),
                 "dtype": str(t.dtype), "device": str(t.device), "원소 수": t.numel()})

pd.DataFrame(rows)
```

## 3. 모양 바꾸기 — 원소 수는 유지됩니다


```python
z = torch.arange(12)
print(z)

print(z.reshape(3, 4))
print(z.reshape(3, 4).shape)
print(z.reshape(2, -1).shape)    # -1 은 알아서 계산
print(z.view(4, 3).shape)        # view 는 메모리를 공유
```

```python
# 원소 수가 맞지 않으면 오류
try:
    z.reshape(5, 3)
except RuntimeError as err:
    print("RuntimeError:", err)
```

## 4. 축 추가와 제거 — 배치 차원을 만들 때 씁니다


```python
v = torch.tensor([1.0, 2.0, 3.0])
print(v.shape)                       # (3,)

print(v.unsqueeze(0).shape)          # (1, 3)  행 방향 추가
print(v.unsqueeze(1).shape)          # (3, 1)  열 방향 추가

w = torch.zeros(1, 3, 1)
print(w.squeeze().shape)             # (3,)  크기 1인 축 제거
```

## 5. 전치와 축 바꾸기


```python
m = torch.randn(2, 5)
print(m.T.shape)                     # (5, 2)

img = torch.randn(32, 32, 3)         # (높이, 너비, 채널)
print(img.permute(2, 0, 1).shape)    # (채널, 높이, 너비) — PyTorch 이미지 순서
```
