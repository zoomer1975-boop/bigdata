## 목표
- 정수 · 실수 · 0 · 난수 Tensor를 만든다
- 자료형을 확인하고 바꾼다
- NumPy 배열과 오간다


```python
import torch
import numpy as np
print("torch", torch.__version__)
```

## 1. 네 가지 생성 방법


```python
a = torch.tensor([1, 2, 3])            # 정수
b = torch.tensor([1.0, 2.0, 3.0])      # 실수
c = torch.zeros(2, 3)                  # 0으로 채우기
d = torch.randn(2, 3)                  # 표준정규분포 난수

print(a, a.dtype)
print(b, b.dtype)
print(c)
print(d)
```

```python
# 그 외 자주 쓰는 것
print(torch.ones(2, 2))
print(torch.arange(0, 10, 2))
print(torch.linspace(0, 1, 5))
print(torch.eye(3))                    # 단위행렬
print(torch.full((2, 2), 7.0))
```

## 2. 자료형 — 딥러닝의 기본은 float32
정수 Tensor로는 미분을 할 수 없습니다. 학습에 쓸 값은 실수여야 합니다.


```python
print(torch.tensor([1, 2, 3]).dtype)          # int64
print(torch.tensor([1.0, 2.0]).dtype)          # float32
print(torch.tensor([True, False]).dtype)       # bool
```

```python
x = torch.tensor([1, 2, 3])
print(x.float().dtype)                 # float32 로
print(x.double().dtype)                # float64 로
print(x.to(torch.float32).dtype)       # 명시적으로

# 만들 때 지정하는 것이 가장 명확합니다
y = torch.tensor([1, 2, 3], dtype=torch.float32)
print(y, y.dtype)
```

## 3. 재현 가능한 난수
같은 seed를 주면 매번 같은 난수가 나옵니다. 실험 결과를 비교할 때 반드시 필요합니다.


```python
torch.manual_seed(42)
print(torch.randn(3))

torch.manual_seed(42)
print(torch.randn(3))     # 같은 값
```

## 4. NumPy와 오가기


```python
arr = np.array([[1.0, 2.0], [3.0, 4.0]])

t = torch.from_numpy(arr)      # NumPy → Tensor
back = t.numpy()               # Tensor → NumPy

print(t.dtype)                 # float64 — NumPy 기본값이 그대로 따라옵니다
print(t.float().dtype)         # 학습 전에는 float32 로 바꿉니다
```

```python
# 주의: from_numpy 는 메모리를 공유합니다
arr[0, 0] = 999
print(t)          # Tensor 도 함께 바뀌었습니다
```

```python
# 복사가 필요하면 torch.tensor 를 씁니다
safe = torch.tensor(arr)
arr[0, 0] = -1
print(safe)       # 영향 없음
```
