## 목표
- CSV를 DataFrame으로 불러온다
- 행·열·자료형·결측값을 한 번에 파악한다
- `head` `info` `describe` `value_counts` 를 읽는다


```python
csv = """name,gender,department,study_hours,attendance,midterm,final
김통계,남,통계학과,12.5,95%,88,92
이확률,여,통계학과,8.0,88%,92,85
박회귀,남,컴퓨터공학과,,72%,79,68
최추정,여,통계학과,15.0,100%,95,98
정검정,남,경제학과,4.5,61%,61,
한분산,여,컴퓨터공학과,10.0,90%,84,88
오평균,남,경제학과,6.5,,70,74
서표본,여,통계학과,13.0,97%,100,96
남표준,남,컴퓨터공학과,9.5,85%,66,71
윤편차,여,경제학과,,80%,82,79
"""

with open("students.csv", "w", encoding="utf-8") as f:
    f.write(csv)

print("students.csv 생성 완료")
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df
```

## 1. 크기와 앞부분 — 제일 먼저 보는 두 가지


```python
print("shape :", df.shape)      # (행, 열)
print("행 개수:", len(df))
print("열 이름:", list(df.columns))

df.head(3)
```

## 2. info() — 자료형과 결측값을 한 화면에

읽는 순서
1. `Non-Null Count` 가 전체 행 수보다 작은 열 → 결측값 있음
2. `Dtype` 이 `object` 인 열 → 숫자가 아님 (계산 불가)


```python
df.info()
```

## 3. describe() — 수치 열의 분포


```python
df.describe()
```

```python
# 문자 열까지 함께 보려면
df.describe(include="all")
```

## 4. 결측값 세기


```python
print(df.isnull().sum())        # 열별 결측 개수
print("\n전체 결측 개수:", df.isnull().sum().sum())
```

## 5. 범주형 값의 분포 — value_counts


```python
print(df["department"].value_counts())
print()
print(df["gender"].value_counts())
```

## 6. 열과 행 고르기


```python
print(df["midterm"].head())            # 한 열 (Series)
print()
print(df[["midterm", "final"]].head())  # 여러 열 (DataFrame)
```

```python
print(df.loc[0])                  # 0번 행 (이름 기준)
print()
print(df.iloc[0:3, 0:3])          # 위치 기준 슬라이싱
```

```python
# 조건으로 고르기 — NumPy의 불리언 인덱싱과 같은 방식
df[df["midterm"] >= 90]
```
