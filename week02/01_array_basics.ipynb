## 목표
- 리스트와 NumPy 배열의 차이를 눈으로 확인한다
- 배열을 여러 방법으로 만든다
- `shape`, `ndim`, `dtype` 을 읽는다


```python
import numpy as np
print("numpy 버전:", np.__version__)
```

## 1. 리스트와 배열은 다르게 동작합니다
같은 `* 2` 가 완전히 다른 결과를 냅니다.


```python
nums_list = [1, 2, 3, 4]
nums_arr = np.array([1, 2, 3, 4])

print("리스트 * 2 :", nums_list * 2)   # 이어 붙이기
print("배열 * 2   :", nums_arr * 2)    # 원소마다 곱하기
```

```python
# 리스트는 원소별 계산에 반복문이 필요합니다
doubled = [n * 2 for n in nums_list]
print(doubled)

# 배열은 한 줄로 끝납니다
print(nums_arr * 2)
print(nums_arr + 100)
print(nums_arr ** 2)
```

## 2. 배열 만들기


```python
a = np.array([1, 2, 3])                 # 리스트에서
b = np.zeros((2, 3))                    # 0으로 채우기
c = np.ones((2, 3))                     # 1로 채우기
d = np.arange(0, 10, 2)                 # 시작·끝·간격
e = np.linspace(0, 1, 5)                # 구간을 5등분
f = np.random.randn(2, 3)               # 표준정규분포

print(a)
print(b)
print(d)
print(e)
print(f)
```

## 3. shape · ndim · dtype
딥러닝 코드에서 가장 자주 확인하는 세 가지입니다.


```python
x = np.array([[1, 2, 3],
              [4, 5, 6]])

print("shape :", x.shape)   # (행, 열)
print("ndim  :", x.ndim)    # 차원 수
print("dtype :", x.dtype)   # 자료형
print("size  :", x.size)    # 전체 원소 수
```

```python
# dtype 바꾸기 — 딥러닝은 보통 float32 를 씁니다
y = x.astype("float32")
print(y.dtype)
print(y)
```

## 4. 차원 늘려 보기


```python
s = np.array(7)                          # 0차원 (스칼라)
v = np.array([1, 2, 3])                  # 1차원
m = np.array([[1, 2], [3, 4]])           # 2차원
t = np.zeros((2, 3, 4))                  # 3차원

for name, arr in [("0차원", s), ("1차원", v), ("2차원", m), ("3차원", t)]:
    print(f"{name}  shape={arr.shape}  ndim={arr.ndim}")
```

## 5. 모양 바꾸기


```python
z = np.arange(12)
print(z)
print(z.reshape(3, 4))
print(z.reshape(3, 4).T)      # 전치
print(z.reshape(2, -1))       # -1 은 알아서 계산
```
