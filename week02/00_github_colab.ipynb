## 3. GitHub에 있는 데이터 파일을 바로 읽기
파일 하나만 필요하면 clone 없이 **raw 주소**로 읽습니다.

```
github.com/<계정>/<저장소>/blob/main/<경로>        ← 사람이 보는 화면
raw.githubusercontent.com/<계정>/<저장소>/main/<경로>  ← 파일 내용 그 자체
```


````python
import pandas as pd

# 예시 주소 — 저장소에 data/scores.csv 를 올린 뒤 실행하세요
url = ("https://raw.githubusercontent.com/"
       "zoomer1975-boop/bigdata/main/data/scores.csv")

try:
    df = pd.read_csv(url)
    print(df.head())
except Exception as err:
    print("아직 파일이 없습니다:", type(err).__name__)
````

## 5. 자주 겪는 문제


````python
# 404 가 뜨는 경우 — 경로나 브랜치 이름이 틀렸습니다
# 확인 순서: 계정 이름 / 저장소 이름 / 브랜치(main) / 폴더 / 파일명 대소문자

import urllib.request

def check(url):
    try:
        urllib.request.urlopen(url)
        return "OK"
    except Exception as e:
        return f"실패 {e}"

print(check("https://raw.githubusercontent.com/zoomer1975-boop/"
            "bigdata/main/week02/00_github_colab.ipynb"))
````
