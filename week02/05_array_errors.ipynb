배열 오류의 해결은 거의 항상 같은 한 줄에서 시작합니다.

```python
print(a.shape, b.shape)
```

아래 셀들은 **일부러 오류가 나도록** 만들어져 있습니다.


````python
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])          # (2, 3)
b = np.array([[1, 2],
              [3, 4]])             # (2, 2)

print(a.shape, b.shape)
````

## 1. IndexError — 없는 위치를 꺼냈다


````python
print(a[2, 0])
````

````python
# 해결: 행은 0 ~ shape[0]-1 까지
print(a.shape)
print(a[1, 0])
````

## 2. ValueError — 크기가 맞지 않는 배열끼리 계산했다


````python
a + b
````

````python
# 해결: 모양을 확인하고 맞춘다
print(a.shape, b.shape)
print(a[:, 0:2] + b)
````

## 3. ValueError — reshape 개수가 맞지 않는다


````python
a.reshape(4, 2)
````

````python
# 해결: 전체 원소 수는 유지되어야 합니다
print("원소 수:", a.size)
print(a.reshape(3, 2))
print(a.reshape(6, -1))
````

## 4. ValueError — 행렬곱의 차원이 맞지 않는다
`(m, n) @ (n, p)` 에서 **가운데 n이 같아야** 합니다.


````python
a @ b
````

````python
# 해결: (2, 3) @ (3, 2) 로 맞춘다
c = np.array([[1, 2],
              [3, 4],
              [5, 6]])
print(a.shape, c.shape)
print(a @ c)
print((a @ c).shape)
````

## 5. 브로드캐스팅이 되는 경우와 안 되는 경우
뒤에서부터 축을 비교해 **같거나 한쪽이 1** 이면 됩니다.


````python
print(a + np.array([10, 20, 30]))          # (2,3) + (3,)   OK
print(a + np.array([[10], [20]]))          # (2,3) + (2,1)  OK
````

````python
a + np.array([10, 20])                     # (2,3) + (2,)   실패
````

````python
# 해결: 열 방향으로 붙이려면 (2, 1) 로 모양을 바꿉니다
print(a + np.array([10, 20]).reshape(2, 1))
````

## 직접 해보기
아래 셀에는 오류가 두 개 있습니다. shape을 찍어 가며 고치세요.


````python
x = np.arange(6).reshape(2, 3)
y = np.arange(6).reshape(2, 3)

print(x @ y)
print(x[0, 3])
````
