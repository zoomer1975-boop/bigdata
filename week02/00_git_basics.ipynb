## 2. 파일을 만들고 상태 확인
`git status` 는 지금 무엇이 기록 대기 중인지 알려 줍니다. 막히면 항상 이 명령부터.

## 3. 스테이징 → 커밋

- `git commit` : 고른 것을 하나의 기록으로 확정한다


```python
!git add analysis.py
!git status
```

```python
!git commit -m "평균 계산 함수 추가"
```

## 4. 두 번째 변경 — diff로 무엇이 달라졌는지 보기


```python
code = '''def mean(values):
    return sum(values) / len(values)


def std(values):
    m = mean(values)
    return (sum((v - m) ** 2 for v in values) / len(values)) ** 0.5
'''
with open("analysis.py", "w") as f:
    f.write(code)

!git diff
```

```python
!git add analysis.py
!git commit -m "표준편차 함수 추가"

!git log --oneline
```

## 5. 되돌리기
- 커밋 전 수정 취소 : `git restore <파일>`
- 스테이징만 취소 : `git restore --staged <파일>`
- 특정 시점의 파일 꺼내기 : `git checkout <커밋> -- <파일>`


```python
# 파일을 망가뜨려 보고
with open("analysis.py", "w") as f:
    f.write("망가진 코드")

print(open("analysis.py").read())
```

```python
# 마지막 커밋 상태로 되돌립니다
!git restore analysis.py

print(open("analysis.py").read())
```

## 7. 브랜치 — 갈라져서 작업하고 합치기


```python
!git switch -c feature-median
!git branch
```

```python
code = open("analysis.py").read() + '''

def median(values):
    s = sorted(values)
    n = len(s)
    return s[n // 2] if n % 2 else (s[n // 2 - 1] + s[n // 2]) / 2
'''
with open("analysis.py", "w") as f:
    f.write(code)

!git add analysis.py
!git commit -m "중앙값 함수 추가"
!git log --oneline
```

```python
# main 으로 돌아가면 중앙값 함수가 없습니다
!git switch main
print(open("analysis.py").read())
```

```python
# 합치기
!git merge feature-median
print(open("analysis.py").read())
!git log --oneline --graph --all
```

## 8. 원격 저장소
지금까지는 모두 **내 컴퓨터 안**의 일입니다. GitHub 같은 원격 저장소와 주고받을 때 쓰는 명령은 세 개입니다.

| 명령 | 하는 일 |
| --- | --- |
| `git clone <주소>` | 원격 저장소를 그대로 내려받는다 |
| `git pull` | 원격의 새 커밋을 내 저장소로 가져온다 |
| `git push` | 내 커밋을 원격으로 올린다 |

`clone` 과 `pull` 은 공개 저장소라면 인증 없이 됩니다. `push` 는 권한이 필요합니다.
