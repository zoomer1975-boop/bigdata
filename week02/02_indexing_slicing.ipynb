## 목표
- 행과 열을 골라낸다
- 슬라이싱으로 부분 배열을 자른다
- 조건으로 값을 골라낸다 (불리언 인덱싱)


```python
import numpy as np

scores = np.array([[88, 92, 79],
                   [95, 61, 84],
                   [70, 100, 66],
                   [82, 74, 91]])

print(scores)
print("shape:", scores.shape)   # 학생 4명 × 과목 3개
```

## 1. 하나 꺼내기 — [행, 열]


```python
print(scores[0, 0])     # 첫 학생의 첫 과목
print(scores[2, 1])     # 세 번째 학생의 두 번째 과목
print(scores[-1, -1])   # 마지막 학생의 마지막 과목
```

## 2. 행 또는 열 전체 — 콜론은 "전부"


```python
print("첫 학생의 모든 점수 :", scores[0, :])
print("첫 과목의 모든 학생 :", scores[:, 0])
print("두 번째 과목        :", scores[:, 1])
```

## 3. 슬라이싱 — 끝 번호는 포함되지 않습니다


```python
print(scores[0:2, :])       # 1~2번 학생
print(scores[:, 1:3])       # 2~3번 과목
print(scores[1:3, 0:2])     # 겹치는 사각형 영역
```

## 4. 불리언 인덱싱 — 조건으로 고르기


```python
print(scores >= 90)              # True/False 배열
print(scores[scores >= 90])      # 조건을 만족하는 값만
print("90점 이상 개수:", (scores >= 90).sum())
```

```python
# 조건을 만족하는 값 바꾸기 — 이미지 처리에서 그대로 쓰입니다
capped = scores.copy()
capped[capped < 70] = 70
print(capped)
```

## 5. copy 를 하지 않으면 원본이 바뀝니다


```python
view = scores[0, :]      # 원본을 들여다보는 창
view[0] = 0
print("원본이 바뀌었습니다:")
print(scores)
```

```python
scores[0, 0] = 88        # 되돌리기
safe = scores[0, :].copy()
safe[0] = 0
print(safe)
print(scores[0, :])      # 원본은 그대로
```
