## 목표
- 흑백 이미지가 2차원 배열임을 확인한다
- 픽셀 값을 직접 바꿔 본다
- 컬러 이미지가 3차원 배열임을 확인한다


```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. 5×5 흑백 이미지 만들기
0은 검정, 255는 흰색입니다.


```python
img = np.zeros((5, 5), dtype=np.uint8)
print(img)
print("shape:", img.shape, "dtype:", img.dtype)
```

```python
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("5x5 all black")
plt.colorbar()
plt.show()
```

## 2. 픽셀 값을 바꾸면 그림이 바뀝니다


```python
img[2, 2] = 255          # 가운데 한 점
img[0, :] = 128          # 첫 행 전체를 회색으로
img[:, 4] = 200          # 마지막 열

print(img)

plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("changed pixels")
plt.show()
```

## 3. 대각선 그리기 — 반복문 없이


```python
diag = np.zeros((9, 9), dtype=np.uint8)
np.fill_diagonal(diag, 255)

plt.imshow(diag, cmap="gray", vmin=0, vmax=255)
plt.title("diagonal")
plt.show()
```

## 4. 숫자 배열과 그림을 나란히 보기


```python
grad = np.arange(0, 256, 16, dtype=np.uint8).reshape(4, 4)
print(grad)

fig, ax = plt.subplots(1, 2, figsize=(7, 3))
ax[0].imshow(grad, cmap="gray", vmin=0, vmax=255)
ax[0].set_title("as image")
ax[1].axis("off")
for i in range(4):
    for j in range(4):
        ax[1].text(j / 4, 1 - i / 4, str(grad[i, j]), fontsize=12)
ax[1].set_title("as numbers")
plt.show()
```

## 5. 컬러 이미지는 3차원 배열입니다
(높이, 너비, 3) — 마지막 축이 빨강·초록·파랑입니다.


```python
color = np.zeros((5, 5, 3), dtype=np.uint8)
color[:, :, 0] = 255          # 빨강 채널만 최대

print("shape:", color.shape)

plt.imshow(color)
plt.title("red")
plt.show()
```

```python
# 채널을 나눠 보기
color[0:2, :, :] = [0, 0, 255]     # 위쪽 두 행은 파랑
color[3:5, :, :] = [0, 255, 0]     # 아래쪽 두 행은 초록

plt.imshow(color)
plt.title("3 channels")
plt.show()

print("빨강 채널만:")
print(color[:, :, 0])
```

## 6. 밝기와 반전 — 배열 연산이 곧 이미지 처리


```python
photo = np.random.randint(0, 256, (8, 8)).astype(np.uint8)

bright = np.clip(photo.astype(int) + 60, 0, 255).astype(np.uint8)
invert = 255 - photo

fig, ax = plt.subplots(1, 3, figsize=(10, 3))
ax[0].imshow(photo, cmap="gray", vmin=0, vmax=255); ax[0].set_title("original")
ax[1].imshow(bright, cmap="gray", vmin=0, vmax=255); ax[1].set_title("+60")
ax[2].imshow(invert, cmap="gray", vmin=0, vmax=255); ax[2].set_title("255 - x")
plt.show()
```
