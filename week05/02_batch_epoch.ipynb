## 목표
- batch size와 반복 횟수의 관계를 계산한다
- epoch이 무엇인지 확인한다
- batch size가 학습에 미치는 영향을 이해한다

데이터 1,000개를 batch size 100으로 학습하면 한 epoch에 10번의 반복이 일어납니다.


```python
import torch
from torch.utils.data import TensorDataset, DataLoader

X = torch.randn(1000, 5)
y = torch.randn(1000, 1)
dataset = TensorDataset(X, y)
print("전체 개수:", len(dataset))
```

## 1. batch size에 따른 반복 횟수


```python
import math

for bs in [1, 10, 100, 250, 1000]:
    steps = math.ceil(len(dataset) / bs)
    print(f"batch_size {bs:5d}  →  한 epoch 당 {steps:5d} 번 반복")
```

```python
# DataLoader 로 확인
for bs in [100, 250, 300]:
    loader = DataLoader(dataset, batch_size=bs)
    print(f"batch_size {bs:4d}  len(loader) = {len(loader)}")
```

## 2. epoch — 전체 데이터를 한 바퀴 도는 것


```python
loader = DataLoader(dataset, batch_size=250, shuffle=True)
epochs = 3

for epoch in range(epochs):
    seen = 0
    for bx, by in loader:
        seen += bx.shape[0]
    print(f"epoch {epoch}  본 데이터 개수 {seen}  반복 {len(loader)}회")
```

## 3. 용어 정리

| 용어 | 뜻 |
| --- | --- |
| batch | 한 번에 모델에 넣는 데이터 묶음 |
| batch size | 그 묶음의 크기 |
| step (iteration) | batch 하나를 처리하는 한 번의 반복 |
| epoch | 전체 데이터를 한 바퀴 다 본 것 |

전체 반복 횟수 = epoch 수 × 한 epoch의 step 수


```python
epochs, bs = 10, 100
steps_per_epoch = math.ceil(1000 / bs)
print("총 반복 횟수:", epochs * steps_per_epoch)
```

## 4. batch size는 무엇을 바꾸는가

| batch size | 메모리 | 갱신 횟수 | 학습 안정성 |
| --- | --- | --- | --- |
| 작다 (8~32) | 적게 씀 | 많다 | 흔들리며 내려감 |
| 크다 (256~) | 많이 씀 | 적다 | 매끄럽게 내려감 |

메모리가 부족하면 batch size를 줄이는 것이 첫 번째 대응입니다. 보통 32, 64, 128을 씁니다.


```python
# 한 batch가 차지하는 메모리 크기 어림 계산
for bs in [32, 128, 512]:
    bytes_ = bs * 5 * 4          # 특성 5개 × float32(4바이트)
    print(f"batch_size {bs:4d}  →  약 {bytes_/1024:.1f} KB")
```
