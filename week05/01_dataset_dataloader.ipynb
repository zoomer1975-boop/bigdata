## 목표
- TensorDataset으로 데이터를 묶는다
- DataLoader로 batch를 꺼낸다
- batch의 shape을 읽는다

Dataset은 "데이터 하나를 꺼내는 방법", DataLoader는 "batch로 묶어 전달하는 방법"입니다.


```python
import torch
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(42)

X = torch.randn(20, 3)              # 데이터 20개, 특성 3개
y = torch.randint(0, 2, (20, 1)).float()

print("X:", X.shape, "y:", y.shape)
```

## 1. Dataset — 하나씩 꺼내기


```python
dataset = TensorDataset(X, y)

print("전체 개수:", len(dataset))

x0, y0 = dataset[0]                 # 0번째 하나
print("한 개:", x0.shape, y0.shape)
print(x0, y0)
```

## 2. DataLoader — batch로 묶어 꺼내기


```python
loader = DataLoader(dataset, batch_size=5, shuffle=True)

print("batch 개수:", len(loader))   # 20 / 5 = 4

for i, (bx, by) in enumerate(loader):
    print(f"batch {i}  X {tuple(bx.shape)}  y {tuple(by.shape)}")
```

## 3. batch의 shape 읽기
`(batch_size, 특성 수)` 입니다. 첫 번째 숫자만 batch에 따라 바뀝니다.


```python
bx, by = next(iter(loader))

print("X batch:", bx.shape)         # (5, 3)
print("y batch:", by.shape)         # (5, 1)
print(bx)
```

## 4. 나누어 떨어지지 않으면 마지막 batch가 작습니다


```python
loader = DataLoader(dataset, batch_size=6, shuffle=False)

for i, (bx, by) in enumerate(loader):
    print(f"batch {i}  크기 {bx.shape[0]}")     # 6, 6, 6, 2
```

```python
# 크기를 맞춰야 할 때는 마지막을 버립니다
loader = DataLoader(dataset, batch_size=6, drop_last=True)
print("batch 개수:", len(loader))               # 3
```

## 5. shuffle의 효과
학습용은 `shuffle=True`, 검증·평가용은 `shuffle=False` 입니다.


```python
simple = TensorDataset(torch.arange(10).float().reshape(-1, 1))

print("shuffle=False")
for (b,) in DataLoader(simple, batch_size=5, shuffle=False):
    print(b.flatten().tolist())

print("\nshuffle=True")
torch.manual_seed(0)
for (b,) in DataLoader(simple, batch_size=5, shuffle=True):
    print(b.flatten().tolist())
```
