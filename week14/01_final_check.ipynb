## 2. 자기 점검 함수
프로젝트 노트북에서 값을 옮겨 적으면 빠진 항목을 알려줍니다.


```python
def check_project(info):
    """info: 프로젝트 정보 딕셔너리"""
    required = {
        "문제": "해결할 문제를 한 문장으로",
        "데이터": "이름과 개수",
        "입력_shape": "예: (batch, 1, 28, 28)",
        "출력_shape": "예: (batch, 10)",
        "모델": "모델 종류",
        "손실함수": "CrossEntropyLoss 또는 MSELoss",
        "성공기준": "숫자로",
        "최종성능": "숫자로",
        "학습시간": "초 단위",
        "주요한계": "숫자 근거와 함께",
    }
    missing = [k for k in required if not info.get(k)]
    print(f"채운 항목 {len(required)-len(missing)} / {len(required)}\n")
    for k, hint in required.items():
        mark = "O" if info.get(k) else "X"
        val = info.get(k, f"({hint})")
        print(f"  [{mark}] {k:12s} {val}")
    if missing:
        print(f"\n빠진 항목: {', '.join(missing)}")
    else:
        print("\n모두 채웠습니다")


# 자기 프로젝트 값으로 채우세요
my_project = {
    "문제": "의류 이미지를 10종류로 분류한다",
    "데이터": "Fashion-MNIST 학습 60,000 / 검증 10,000",
    "입력_shape": "(batch, 1, 28, 28)",
    "출력_shape": "(batch, 10)",
    "모델": "CNN (Conv 2층 + BatchNorm + Dropout)",
    "손실함수": "CrossEntropyLoss",
    "성공기준": "검증 정확도 0.88",
    "최종성능": "",
    "학습시간": "",
    "주요한계": "",
}
check_project(my_project)
```

## 3. 발표용 그림 한 번에 뽑기
학습이 끝난 상태에서 아래 셀을 돌리면 발표 자료 4·5장에 쓸 그림이 나옵니다.


```python
import matplotlib.pyplot as plt
import numpy as np

def presentation_figures(history, save=True):
    """history: [(train_loss, train_acc, val_loss, val_acc), ...]"""
    tr_l = [h[0] for h in history]; tr_a = [h[1] for h in history]
    va_l = [h[2] for h in history]; va_a = [h[3] for h in history]
    xs = range(1, len(history)+1)

    fig, ax = plt.subplots(1, 2, figsize=(12, 4))
    ax[0].plot(xs, tr_l, label="학습", linewidth=2)
    ax[0].plot(xs, va_l, label="검증", linewidth=2)
    ax[0].set_title("Loss", fontsize=14); ax[0].set_xlabel("epoch")
    ax[0].legend(fontsize=11); ax[0].grid(alpha=.3)
    ax[1].plot(xs, tr_a, label="학습", linewidth=2)
    ax[1].plot(xs, va_a, label="검증", linewidth=2)
    ax[1].set_title("Accuracy", fontsize=14); ax[1].set_xlabel("epoch")
    ax[1].legend(fontsize=11); ax[1].grid(alpha=.3)
    plt.tight_layout()
    if save:
        plt.savefig("fig_curves.png", dpi=150, bbox_inches="tight")
        print("fig_curves.png 저장")
    plt.show()

    print(f"\n발표 4장에 넣을 숫자")
    print(f"  최종 검증 정확도 {va_a[-1]:.4f}")
    print(f"  최고 검증 정확도 {max(va_a):.4f} (epoch {int(np.argmax(va_a))+1})")
    print(f"  학습-검증 차이   {tr_a[-1]-va_a[-1]:.4f}")


# 예시 history — 자기 것으로 바꾸세요
fake = [(0.9-i*0.09, 0.55+i*0.045, 0.85-i*0.06, 0.58+i*0.038) for i in range(8)]
presentation_figures(fake)
```

저장된 PNG를 PowerPoint에 그대로 붙이면 됩니다. 발표용은 `dpi=150` 이상으로 저장하세요.
