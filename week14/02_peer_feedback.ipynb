## 네 가지 질문

발표마다 아래 네 가지를 적습니다.

1. 가장 인상적인 부분은 무엇인가?
2. 이해하기 어려웠던 부분은 무엇인가?
3. 추가로 궁금한 점은 무엇인가?
4. 개선 아이디어는 무엇인가?


```python
import pandas as pd

# 발표를 들으며 채우세요
feedback = []

def add(team, impressive, unclear, question, idea):
    feedback.append({
        "팀": team,
        "인상적인 부분": impressive,
        "어려웠던 부분": unclear,
        "궁금한 점": question,
        "개선 아이디어": idea,
    })


# 예시
add("1팀",
    "혼동행렬로 셔츠와 코트가 헷갈리는 것을 짚은 부분",
    "데이터 증강을 왜 그 조합으로 골랐는지",
    "증강 없이 학습하면 차이가 얼마나 나는가",
    "클래스 가중치를 주면 셔츠 정확도가 오를 것 같다")

# 여기에 계속 추가하세요
# add("2팀", "", "", "", "")

pd.DataFrame(feedback)
```
