## 7. 배운 것을 코드로 확인 — 짧은 복습
아래 셀들은 각 주차의 핵심을 한 줄씩 다시 실행합니다.


````python
# 4주차 — Tensor와 자동미분
x = torch.tensor(3.0, requires_grad=True)
(x ** 2).backward()
print("4주차  dy/dx =", x.grad.item())

# 6주차 — 계층
m = nn.Sequential(nn.Linear(4, 8), nn.ReLU(), nn.Linear(8, 3))
print("6주차  파라미터", sum(p.numel() for p in m.parameters()))

# 8주차 — 합성곱과 풀링
import torch.nn.functional as F
img = torch.randn(1, 1, 28, 28)
print("8주차  28x28 → Conv+Pool →", tuple(F.max_pool2d(F.conv2d(img, torch.randn(16,1,3,3), padding=1), 2).shape))

# 11주차 — LSTM
out, _ = nn.LSTM(1, 16, batch_first=True)(torch.randn(4, 20, 1))
print("11주차 LSTM 출력", tuple(out.shape))

# 12주차 — 임베딩
print("12주차 임베딩 출력", tuple(nn.Embedding(1000, 64)(torch.tensor([[5, 12, 0]])).shape))
````

## 8. 앞으로의 학습 방향

| 방향 | 무엇인가 | 오늘까지 배운 것과의 연결 |
| --- | --- | --- |
| Transformer | 순서를 은닉 상태 없이 다루는 구조 | 12주차 임베딩 다음 단계 |
| 객체 탐지 | 이미지에서 위치까지 찾는 문제 | 8주차 CNN의 확장 |
| 이미지 생성 | 없는 이미지를 만들어내는 모델 | 새 영역 |
| 모델 배포 | 학습한 모델을 서비스로 | `state_dict` 저장 다음 단계 |
| 대규모 언어 모델 | 문장을 이해하고 생성하는 큰 모델 | 12주차 자연어 처리의 연장 |
